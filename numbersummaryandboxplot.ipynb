```python
import numpy as np
lst_marks=[45,78,98,45,67,34,56,78,67,90,32,46,87,78]
minimum,Q1,median,Q3,maximum=np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*(IQR)
higher_fence=Q3+1.5*(IQR)
```

```python
print(lower_fence)
print(higher_fence)
```
